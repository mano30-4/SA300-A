```python
class Item:
  discount = 100

item1 = Item()
item2 = Item()
## before updation
print(item1.discount)
print(item2.discount)
Item.discount = 1
print(item1.discount)
print(item2.discount)
```

```python
class Item:
  Name = None
  price = 0
  quantity = 0
# we have a class and class variables every object can access which makes a own of copy variable for the particular object and we can update it


item1= Item()
print(item1)
print(vars(item1)) # we created object but no variables are assiocated with it

item1.Name="phone" ##item1 has variable Name and its value is changed to "phone"
item1.price=1000.  ## Same as above step
item1.quantity=1   ## Same as above step
print("\n\n",vars(item1))
```

```python
## __init__
class Item():
  def __init__(self):
    self.Name=None
    self.price=0
    self.quantity=0
  ## 'self' refers to object we dealing with, it gets passed as a default argument for the func and we assigning values to object variables which are get created everytime object is created (bc of __init__)
item1=Item()
print(vars(item1))

```

```python
#contitnuation for above , just th way changed how we assign values
#earlier we constanly assigning constant values to object variable when ever object gets created
#now we getting custom values for object variables from user in arguments and assigning them to object variables when object gets created itself
class Item():
  def __init__(self,name:str, price:int, quantity :int) :
    self.Name=name
    self.price=price
    self.quantity=quantity

item1=Item("phone","1000","10")
print(vars(item1))
```

```python
#assert statement (assert 'boolean statement','exception messages')
#helps u for input validation for int arguments
class Item():
  def __init__(self,name:str, price:float, quantity :int) :
    assert price>=0, "price can't be negative" #try giving negative inputs
    assert quantity>0, "quantity can't be negative"
    self.Name=name
    self.price=price
    self.quantity=quantity

item1=Item("phone",1000,10) # works fine
item2=Item("tablet",-1000,-10) ## breaks
```

```python
# we are trying to make list of all objects which are getting created
class Item():

  allItems=[] ## class Variable, a list to stores objects

  def __init__(self,name:str, price:float, quantity :int) :
    assert price>=0, "price can't be negative"
    assert quantity>0, "quantity can't be negative"
    self.name=name
    self.price=price
    self.quantity=quantity

    Item.allItems.append(self) # 'Item.' alias 'ClassName.' says we are referencing allitems which is Class Item
                              # if we use allitems. python wil search for it in funtion scope(__init__) then it will go for main class (global scope) but our list(variable) is in class Named 'Item' so we need to reference it so python searches there

item1=Item("phone",1000,10)
item2=Item("tablet",500,5)
item3=Item("laptop",10000,1)

print(Item.allItems)
print(Item.allItems[0])
print(Item.allItems[0].name)
```

```python
##__repr__ helps how to represent object when we use print use 'object' in print
## it gets called only when there is object in print statement
## now we will use same code as above and represent objects in custom string format

# we are trying to make list of all objects which are getting created
class Item():

  allItems=[] ## class Variable, a list to stores objects

  def __init__(self,name:str, price:float, quantity :int) :
    assert price>=0, "price can't be negative"
    assert quantity>0, "quantity can't be negative"
    self.name=name
    self.price=price
    self.quantity=quantity

    Item.allItems.append(self) # 'Item.' alias 'ClassName.' says we are referencing allitems which is Class Item
                              # if we use allitems. python wil search for it in funtion scope(__init__) then it will go for main class (global scope) but our list(variable) is in class Named 'Item' so we need to reference it so python searches there
  def __repr__(self):
    return f"Item('{self.name}',{self.price},{self.quantity})"


item1=Item("phone",1000,10)
item2=Item("tablet",500,5)
item3=Item("laptop",10000,1)

print(Item.allItems)
print(Item.allItems[0])
```
